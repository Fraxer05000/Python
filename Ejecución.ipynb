```python
pdf_analysis_code = r'''import hashlib
import re
import unicodedata
from dataclasses import dataclass
from decimal import Decimal

import pytesseract
from pdf2image import convert_from_path
from pypdf import PdfReader
from django.db import transaction
from django.utils import timezone


STOPWORDS = {
    "de", "del", "la", "las", "el", "los", "y", "en", "a", "al", "por",
    "para", "con", "sin", "sobre", "segun", "bajo", "un", "una", "uno",
    "numero", "tiempo", "presentados", "reportada", "reportado", "indicador",
    "indicadores", "poa",
}

PALABRAS_CONGELAR = [
    "suspension",
    "suspension temporal",
    "congelamiento",
    "congelar",
    "suspender",
    "impiden el cumplimiento",
    "impedimento",
    "no puede cumplirse",
]

PALABRAS_REANUDAR = [
    "reanudacion",
    "reactivacion",
    "reanudar",
    "reactivar",
    "reanudado",
    "reactivado",
]

PALABRAS_SOPORTE_FORMAL = [
    "memorando",
    "oficio",
    "solicitud",
    "poa",
]

CUSTOM_KEYWORDS = {
    "numero de informes consolidados de informacion financiera reportada a tiempo en el e sigef": [
        "informes consolidados",
        "informacion financiera",
        "financiera",
        "e sigef",
        "sigef",
        "reporte financiero",
        "presupuesto",
        "contable",
    ],
    "numero de formularios de declaracion de impuestos presentados a tiempo al sri": [
        "formularios",
        "declaracion de impuestos",
        "impuestos",
        "sri",
        "tributario",
        "tributaria",
        "iva",
        "retencion",
    ],
}


@dataclass
class AnalysisResult:
    accion_detectada: str
    score_relacion: Decimal
    score_accion: Decimal
    estado_resultante: str
    observacion: str
    mensaje: str
    texto_extraido: str
    hash_archivo: str


def normalizar_texto(texto: str) -> str:
    texto = texto.lower()
    texto = unicodedata.normalize("NFKD", texto)
    texto = "".join(c for c in texto if not unicodedata.combining(c))
    texto = re.sub(r"[^a-z0-9\s]", " ", texto)
    texto = re.sub(r"\s+", " ", texto).strip()
    return texto


def extraer_texto_pdf(path_pdf: str) -> str:
    try:
        reader = PdfReader(path_pdf)
        partes = []
        for page in reader.pages:
            partes.append(page.extract_text() or "")
        return "\n".join(partes).strip()
    except Exception:
        return ""


def extraer_texto_ocr(path_pdf: str) -> str:
    try:
        imagenes = convert_from_path(path_pdf, dpi=250)
        partes = []
        for img in imagenes:
            texto = pytesseract.image_to_string(img, lang="spa")
            partes.append(texto)
        return "\n".join(partes).strip()
    except Exception:
        return ""


def calcular_hash_archivo(path_pdf: str) -> str:
    sha256 = hashlib.sha256()
    with open(path_pdf, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            sha256.update(chunk)
    return sha256.hexdigest()


def construir_keywords_indicador(indicador) -> list[str]:
    nombre_normalizado = normalizar_texto(indicador.nombre)

    if nombre_normalizado in CUSTOM_KEYWORDS:
        return CUSTOM_KEYWORDS[nombre_normalizado]

    palabras = []
    for token in nombre_normalizado.split():
        if len(token) >= 4 and token not in STOPWORDS:
            palabras.append(token)

    seen = set()
    resultado = []
    for p in palabras:
        if p not in seen:
            seen.add(p)
            resultado.append(p)

    return resultado[:12]


def calcular_relacion_indicador(texto_normalizado: str, indicador) -> Decimal:
    keywords = construir_keywords_indicador(indicador)

    if not keywords:
        return Decimal("0.00")

    coincidencias = 0
    for kw in keywords:
        if kw in texto_normalizado:
            coincidencias += 1

    score = (coincidencias / len(keywords)) * 100
    return Decimal(f"{score:.2f}")


def detectar_accion(texto_normalizado: str) -> tuple[str, Decimal]:
    score_congelar = sum(1 for kw in PALABRAS_CONGELAR if kw in texto_normalizado)
    score_reanudar = sum(1 for kw in PALABRAS_REANUDAR if kw in texto_normalizado)

    if score_congelar > score_reanudar and score_congelar > 0:
        return "CONGELAR", Decimal(f"{(score_congelar / max(len(PALABRAS_CONGELAR), 1)) * 100:.2f}")

    if score_reanudar > score_congelar and score_reanudar > 0:
        return "REANUDAR", Decimal(f"{(score_reanudar / max(len(PALABRAS_REANUDAR), 1)) * 100:.2f}")

    return "NO_DETERMINADA", Decimal("0.00")


def validar_soporte_formal(texto_normalizado: str) -> bool:
    encontrados = sum(1 for kw in PALABRAS_SOPORTE_FORMAL if kw in texto_normalizado)
    return encontrados >= 2


def resolver_estado(score_relacion: Decimal, accion: str, soporte_formal: bool):
    if score_relacion < Decimal("25.00"):
        return (
            "REQUIERE_REVISION",
            "Baja coincidencia entre el contenido del PDF y el indicador seleccionado.",
            "El documento fue cargado, pero requiere revisión manual por baja coincidencia con el indicador.",
        )

    if not soporte_formal:
        return (
            "REQUIERE_REVISION",
            "El documento no presenta suficiente estructura documental formal para aprobar el cambio automático.",
            "El PDF fue leído, pero requiere validación manual por falta de soporte formal suficiente.",
        )

    if accion == "CONGELAR":
        return (
            "CONGELADO",
            "Documento compatible con suspensión/congelamiento del indicador.",
            "Indicador congelado correctamente con respaldo documental.",
        )

    if accion == "REANUDAR":
        return (
            "REANUDADO",
            "Documento compatible con reanudación/reactivación del indicador.",
            "Indicador reanudado correctamente con respaldo documental.",
        )

    return (
        "REQUIERE_REVISION",
        "No se detectó una acción clara de congelamiento o reanudación.",
        "El documento fue registrado, pero requiere validación manual.",
    )


@transaction.atomic
def analizar_y_actualizar_indicador(indicador, anexo):
    path_pdf = anexo.archivo_pdf.path

    texto = extraer_texto_pdf(path_pdf)

    if len(texto.strip()) < 80:
        texto = extraer_texto_ocr(path_pdf)

    if len(texto.strip()) < 50:
        anexo.texto_extraido = texto
        anexo.accion_detectada = "NO_DETERMINADA"
        anexo.score_relacion = Decimal("0.00")
        anexo.score_accion = Decimal("0.00")
        anexo.estado_resultante = "REQUIERE_REVISION"
        anexo.observacion = "No fue posible extraer suficiente texto del PDF."
        anexo.hash_archivo = calcular_hash_archivo(path_pdf)
        anexo.save()

        indicador.estado_control = "REQUIERE_REVISION"
        indicador.fecha_estado = timezone.now()
        indicador.ultimo_anexo_estado = anexo
        indicador.save(update_fields=["estado_control", "fecha_estado", "ultimo_anexo_estado"])

        return AnalysisResult(
            accion_detectada="NO_DETERMINADA",
            score_relacion=Decimal("0.00"),
            score_accion=Decimal("0.00"),
            estado_resultante="REQUIERE_REVISION",
            observacion="No fue posible extraer suficiente texto del PDF.",
            mensaje="No se pudo leer correctamente el documento. Revise la calidad del escaneo.",
            texto_extraido=texto,
            hash_archivo=anexo.hash_archivo,
        )

    texto_normalizado = normalizar_texto(texto)
    score_relacion = calcular_relacion_indicador(texto_normalizado, indicador)
    accion_detectada, score_accion = detectar_accion(texto_normalizado)
    soporte_formal = validar_soporte_formal(texto_normalizado)

    estado_resultante, observacion, mensaje = resolver_estado(
        score_relacion=score_relacion,
        accion=accion_detectada,
        soporte_formal=soporte_formal,
    )

    anexo.texto_extraido = texto
    anexo.accion_detectada = accion_detectada
    anexo.score_relacion = score_relacion
    anexo.score_accion = score_accion
    anexo.estado_resultante = estado_resultante
    anexo.observacion = observacion
    anexo.hash_archivo = calcular_hash_archivo(path_pdf)
    anexo.save()

    indicador.estado_control = estado_resultante
    indicador.fecha_estado = timezone.now()
    indicador.ultimo_anexo_estado = anexo
    indicador.save(update_fields=["estado_control", "fecha_estado", "ultimo_anexo_estado"])

    return AnalysisResult(
        accion_detectada=accion_detectada,
        score_relacion=score_relacion,
        score_accion=score_accion,
        estado_resultante=estado_resultante,
        observacion=observacion,
        mensaje=mensaje,
        texto_extraido=texto,
        hash_archivo=anexo.hash_archivo,
    )
'''

with open(BASE_DIR / "poa" / "services" / "pdf_analysis.py", "w", encoding="utf-8") as f:
    f.write(pdf_analysis_code)

print("pdf_analysis.py creado.")
```

```python
models_code = r'''from django.conf import settings
from django.db import models
from django.utils import timezone


class Departamento(models.Model):
    nombre = models.CharField(max_length=200, unique=True)

    class Meta:
        ordering = ["nombre"]

    def __str__(self):
        return self.nombre


class Area(models.Model):
    departamento = models.ForeignKey(
        Departamento,
        on_delete=models.CASCADE,
        related_name="areas",
    )
    nombre = models.CharField(max_length=200)

    class Meta:
        ordering = ["nombre"]
        unique_together = ("departamento", "nombre")

    def __str__(self):
        return f"{self.departamento.nombre} - {self.nombre}"


class Indicador(models.Model):
    class EstadoControl(models.TextChoices):
        ACTIVO = "ACTIVO", "Activo"
        CONGELADO = "CONGELADO", "Congelado"
        REANUDADO = "REANUDADO", "Reanudado"
        REQUIERE_REVISION = "REQUIERE_REVISION", "Requiere revisión"

    departamento = models.ForeignKey(
        Departamento,
        on_delete=models.CASCADE,
        related_name="indicadores",
    )
    area = models.ForeignKey(
        Area,
        on_delete=models.SET_NULL,
        null=True,
        blank=True,
        related_name="indicadores",
    )
    codigo = models.CharField(max_length=50, blank=True, null=True)
    nombre = models.CharField(max_length=500)
    descripcion = models.TextField(blank=True, null=True)

    estado_control = models.CharField(
        max_length=30,
        choices=EstadoControl.choices,
        default=EstadoControl.ACTIVO,
    )
    fecha_estado = models.DateTimeField(blank=True, null=True)
    ultimo_anexo_estado = models.ForeignKey(
        "IndicadorAnexo",
        on_delete=models.SET_NULL,
        null=True,
        blank=True,
        related_name="+",
    )

    class Meta:
        ordering = ["nombre"]

    def __str__(self):
        return self.nombre


class IndicadorAnexo(models.Model):
    class AccionDetectada(models.TextChoices):
        CONGELAR = "CONGELAR", "Congelar"
        REANUDAR = "REANUDAR", "Reanudar"
        NO_DETERMINADA = "NO_DETERMINADA", "No determinada"

    class EstadoResultante(models.TextChoices):
        ACTIVO = "ACTIVO", "Activo"
        CONGELADO = "CONGELADO", "Congelado"
        REANUDADO = "REANUDADO", "Reanudado"
        REQUIERE_REVISION = "REQUIERE_REVISION", "Requiere revisión"

    indicador = models.ForeignKey(
        Indicador,
        on_delete=models.CASCADE,
        related_name="anexos",
    )
    departamento = models.ForeignKey(
        Departamento,
        on_delete=models.CASCADE,
        related_name="anexos_indicadores",
    )
    area = models.ForeignKey(
        Area,
        on_delete=models.SET_NULL,
        null=True,
        blank=True,
        related_name="anexos_indicadores",
    )

    archivo_pdf = models.FileField(upload_to="poa/anexos/%Y/%m/")
    nombre_original = models.CharField(max_length=255)

    hash_archivo = models.CharField(max_length=64, blank=True)
    texto_extraido = models.TextField(blank=True)

    accion_detectada = models.CharField(
        max_length=20,
        choices=AccionDetectada.choices,
        default=AccionDetectada.NO_DETERMINADA,
    )
    score_relacion = models.DecimalField(max_digits=5, decimal_places=2, default=0)
    score_accion = models.DecimalField(max_digits=5, decimal_places=2, default=0)

    estado_resultante = models.CharField(
        max_length=30,
        choices=EstadoResultante.choices,
        default=EstadoResultante.REQUIERE_REVISION,
    )
    observacion = models.TextField(blank=True)

    usuario = models.ForeignKey(
        settings.AUTH_USER_MODEL,
        on_delete=models.SET_NULL,
        null=True,
        blank=True,
        related_name="anexos_poa_subidos",
    )
    fecha_registro = models.DateTimeField(default=timezone.now)

    class Meta:
        ordering = ["-fecha_registro"]

    def __str__(self):
        return f"Anexo #{self.pk} - {self.indicador.nombre}"
'''

with open(BASE_DIR / "poa" / "models.py", "w", encoding="utf-8") as f:
    f.write(models_code)

print("models.py creado.")
```

```python
forms_code = r'''from django import forms
from .models import Departamento, Area


class FiltroConfiguracionPOAForm(forms.Form):
    departamento = forms.ModelChoiceField(
        queryset=Departamento.objects.all(),
        required=True,
        empty_label="Seleccione departamento",
    )
    area = forms.ModelChoiceField(
        queryset=Area.objects.none(),
        required=False,
        empty_label="Todas las áreas",
    )

    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)

        departamento_id = None

        if "departamento" in self.data:
            try:
                departamento_id = int(self.data.get("departamento"))
            except (TypeError, ValueError):
                departamento_id = None
        elif self.initial.get("departamento"):
            departamento_id = self.initial["departamento"].id

        if departamento_id:
            self.fields["area"].queryset = Area.objects.filter(
                departamento_id=departamento_id
            ).order_by("nombre")


class AnexoIndicadorForm(forms.Form):
    archivo_pdf = forms.FileField()

    def clean_archivo_pdf(self):
        archivo = self.cleaned_data["archivo_pdf"]

        nombre = archivo.name.lower()
        if not nombre.endswith(".pdf"):
            raise forms.ValidationError("Solo se permiten archivos en formato .pdf")

        content_type = getattr(archivo, "content_type", "")
        if content_type and content_type != "application/pdf":
            raise forms.ValidationError("El archivo cargado no es un PDF válido.")

        if archivo.size > 15 * 1024 * 1024:
            raise forms.ValidationError("El PDF no puede superar 15 MB.")

        return archivo
'''

with open(BASE_DIR / "poa" / "forms.py", "w", encoding="utf-8") as f:
    f.write(forms_code)

print("forms.py creado.")
```

```python
views_code = r'''from django.contrib import messages
from django.contrib.auth.decorators import login_required
from django.shortcuts import get_object_or_404, redirect, render

from .forms import FiltroConfiguracionPOAForm, AnexoIndicadorForm
from .models import Indicador, IndicadorAnexo
from .services.pdf_analysis import analizar_y_actualizar_indicador


@login_required
def configuracion_arbol_poa(request):
    departamento_id = request.GET.get("departamento")
    area_id = request.GET.get("area")

    filtro_form = FiltroConfiguracionPOAForm(request.GET or None)
    upload_form = AnexoIndicadorForm()

    indicadores = Indicador.objects.select_related("departamento", "area").none()

    if filtro_form.is_valid():
        departamento = filtro_form.cleaned_data["departamento"]
        area = filtro_form.cleaned_data["area"]

        indicadores = Indicador.objects.select_related("departamento", "area").filter(
            departamento=departamento
        )

        if area:
            indicadores = indicadores.filter(area=area)

    if request.method == "POST":
        indicador_id = request.POST.get("indicador_id")
        indicador = get_object_or_404(
            Indicador.objects.select_related("departamento", "area"),
            pk=indicador_id
        )

        upload_form = AnexoIndicadorForm(request.POST, request.FILES)

        if upload_form.is_valid():
            archivo = upload_form.cleaned_data["archivo_pdf"]

            anexo = IndicadorAnexo.objects.create(
                indicador=indicador,
                departamento=indicador.departamento,
                area=indicador.area,
                archivo_pdf=archivo,
                nombre_original=archivo.name,
                usuario=request.user,
            )

            resultado = analizar_y_actualizar_indicador(indicador, anexo)

            if resultado.estado_resultante in ["CONGELADO", "REANUDADO"]:
                messages.success(request, resultado.mensaje)
            else:
                messages.warning(request, resultado.mensaje)

            params = f"?departamento={indicador.departamento_id}"
            if indicador.area_id:
                params += f"&area={indicador.area_id}"
            return redirect(f"{request.path}{params}")
        else:
            messages.error(request, "El archivo no es válido. Solo se permite PDF.")

    context = {
        "filtro_form": filtro_form,
        "upload_form": upload_form,
        "indicadores": indicadores,
        "departamento_id": departamento_id,
        "area_id": area_id,
    }
    return render(request, "poa/configuracion_arbol_poa.html", context)
'''

with open(BASE_DIR / "poa" / "views.py", "w", encoding="utf-8") as f:
    f.write(views_code)

print("views.py creado.")
```

```python
import os
from pathlib import Path

# 0) Permitir ORM sync dentro de Jupyter (async context)
os.environ["DJANGO_ALLOW_ASYNC_UNSAFE"] = "true"

# 1) Ruta del proyecto
BASE_DIR = Path(r"C:\Users\Simpl\Downloads\Hector\Python Django")
os.chdir(BASE_DIR)

# 2) Django setup
os.environ.setdefault("DJANGO_SETTINGS_MODULE", "mi_proyecto.settings")
import django
django.setup()

from poa.models import Departamento, Area, Indicador

# 3) Leer Excel (usa el que subiste o el que esté en carpeta)
import pandas as pd

xlsx_path = BASE_DIR / "POA EP-PI 2026 (1).xlsx"
if not xlsx_path.exists():
    # intenta encontrar cualquier xlsx con "POA" en nombre
    candidates = list(BASE_DIR.glob("*.xlsx"))
    xlsx_path = next((p for p in candidates if "POA" in p.name.upper()), None)

if xlsx_path and xlsx_path.exists():
    raw = pd.read_excel(xlsx_path, sheet_name=0, header=None)

    # Cabeceras en fila 2 y 3; datos desde fila 4 (según tu formato)
    hdr1 = raw.iloc[2]
    hdr2 = raw.iloc[3]

    cols = []
    for i in range(raw.shape[1]):
        h2 = hdr2[i]
        h1 = hdr1[i]
        if pd.notna(h2):
            cols.append(str(h2).strip())
        elif pd.notna(h1):
            cols.append(str(h1).strip())
        else:
            cols.append(f"col_{i}")

    data = raw.iloc[4:].copy()
    data.columns = cols

    needed = ["Unidad Administrativa", "Área", "Indicador", "Descripción Indicador"]
    missing = [c for c in needed if c not in data.columns]
    if missing:
        raise ValueError(f"Faltan columnas en el Excel: {missing}")

    df = data[needed].copy()
    df["Unidad Administrativa"] = df["Unidad Administrativa"].ffill()
    df["Área"] = df["Área"].fillna("GENERAL")
    df = df.dropna(subset=["Unidad Administrativa", "Indicador"])

    df["Unidad Administrativa"] = df["Unidad Administrativa"].astype(str).str.strip()
    df["Área"] = df["Área"].astype(str).str.strip().replace({"": "GENERAL"})
    df["Indicador"] = df["Indicador"].astype(str).str.strip()
    df["Descripción Indicador"] = df["Descripción Indicador"].fillna("").astype(str).str.strip()

    df = df[(df["Unidad Administrativa"] != "") & (df["Indicador"] != "") & (df["Indicador"].str.lower() != "nan")]

    print(f"Excel detectado: {xlsx_path.name} | Filas útiles: {len(df)}")
else:
    # Fallback mínimo
    df = pd.DataFrame([
        {
            "Unidad Administrativa": "DIRECCIÓN FINANCIERA",
            "Área": "GENERAL",
            "Indicador": "Número de informes consolidados de información financiera reportada a tiempo en el E-sigef",
            "Descripción Indicador": "Mide el número de informes consolidados de información financiera reportados a tiempo en el E-sigef."
        },
        {
            "Unidad Administrativa": "DIRECCIÓN FINANCIERA",
            "Área": "GENERAL",
            "Indicador": "Número de formularios de declaración de impuestos presentados a tiempo al SRI",
            "Descripción Indicador": "Mide el número de formularios de declaración de impuestos presentados a tiempo al SRI."
        },
    ])
    print("Excel NO encontrado. Se creó dataset mínimo.")

# 4) Seed idempotente (sin transaction.atomic para evitar choque async)
created_dep = 0
created_area = 0
created_ind = 0

dep_cache = {}
area_cache = {}

# Departamentos
for dep_name in df["Unidad Administrativa"].dropna().unique():
    dep_name = str(dep_name).strip()
    if not dep_name:
        continue
    dep_obj, was_created = Departamento.objects.get_or_create(nombre=dep_name)
    dep_cache[dep_name] = dep_obj
    created_dep += int(was_created)

# Áreas
pairs = df[["Unidad Administrativa", "Área"]].dropna().drop_duplicates()
for dep_name, area_name in pairs.itertuples(index=False):
    dep_name = str(dep_name).strip()
    area_name = str(area_name).strip() if area_name else "GENERAL"
    if not area_name:
        area_name = "GENERAL"
    dep_obj = dep_cache[dep_name]
    key = (dep_obj.id, area_name)
    area_obj, was_created = Area.objects.get_or_create(departamento=dep_obj, nombre=area_name)
    area_cache[key] = area_obj
    created_area += int(was_created)

# Indicadores
for dep_name, area_name, ind_name, ind_desc in df.itertuples(index=False):
    dep_name = str(dep_name).strip()
    area_name = str(area_name).strip() if area_name else "GENERAL"
    if not area_name:
        area_name = "GENERAL"
    ind_name = str(ind_name).strip()
    ind_desc = str(ind_desc).strip() if ind_desc else ""

    dep_obj = dep_cache[dep_name]
    area_obj = area_cache[(dep_obj.id, area_name)]

    # Evitar duplicados por (departamento, área, nombre)
    obj, was_created = Indicador.objects.get_or_create(
        departamento=dep_obj,
        area=area_obj,
        nombre=ind_name,
        defaults={"descripcion": ind_desc}
    )
    if was_created:
        created_ind += 1
    else:
        # si estaba vacío y ahora viene con descripción, actualiza
        if (not obj.descripcion) and ind_desc:
            obj.descripcion = ind_desc
            obj.save(update_fields=["descripcion"])

print("✅ Seed completado")
print(f"Departamentos creados: {created_dep}")
print(f"Áreas creadas:         {created_area}")
print(f"Indicadores creados:   {created_ind}")

print("\nTotales actuales en BD:")
print("Departamentos:", Departamento.objects.count())
print("Áreas:", Area.objects.count())
print("Indicadores:", Indicador.objects.count())
```
